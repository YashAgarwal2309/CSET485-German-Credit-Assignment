```python
# ==============================================================================
# CSET485 – AI and Society · Assignment #1 · German Credit Data
# Roll number S24CSEU0244  →  last digit 4 → German Credit ; last 3 digits → random_state 244
# Google-Colab-ready: paste the whole file into one cell (or split at the "CELL" banners) and run.
# Every number is computed from the data; nothing is hard-coded.
# ==============================================================================

# ------------------------------------------------------------------------------
# CELL 1 · Setup: packages, imports, constants, seeds
# ------------------------------------------------------------------------------
import subprocess, sys
for pkg in ["statsmodels", "openpyxl", "seaborn"]:
    try:
        __import__(pkg)
    except ImportError:
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", pkg], check=False)

import os, io, json, warnings, urllib.request
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (confusion_matrix, accuracy_score, precision_score, recall_score,
                             f1_score, roc_auc_score, mean_squared_error)

warnings.filterwarnings("ignore")
pd.set_option("display.width", 180, "display.max_columns", 80, "display.max_rows", 200)
sns.set_theme(style="whitegrid")

ROLL_NUMBER   = "S24CSEU0244"
RANDOM_STATE  = int(ROLL_NUMBER[-3:])      # 244 – required for the 70/30 train-test split
SAMPLING_SEED = RANDOM_STATE               # documented seed for the (conditional) adversarial sampling
TEST_SIZE     = 0.30
np.random.seed(RANDOM_STATE)

OUT_DIR = "outputs"; TAB_DIR = f"{OUT_DIR}/tables"; FIG_DIR = f"{OUT_DIR}/figures"
for d in (OUT_DIR, TAB_DIR, FIG_DIR):
    os.makedirs(d, exist_ok=True)

SUMMARY = {}                                # every headline number goes here → final_results_summary.csv
def record(key, value):
    if isinstance(value, (np.floating, np.integer)):
        value = value.item()
    SUMMARY[key] = value
    return value

def show(title, obj):
    print("\n" + "=" * 100 + f"\n{title}\n" + "=" * 100)
    try:
        display(obj)          # notebook / Colab
    except NameError:
        print(obj)

print(f"Roll number: {ROLL_NUMBER} | dataset: German Credit | random_state: {RANDOM_STATE} | test_size: {TEST_SIZE}")

# ------------------------------------------------------------------------------
# CELL 2 · Load the dataset (UCI Statlog german.data; 1000 rows × 20 attributes + class)
# ------------------------------------------------------------------------------
UCI_URL    = "https://archive.ics.uci.edu/ml/machine-learning-databases/statlog/german/german.data"
MIRROR_URL = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/german.csv"   # identical content, comma separated
LOCAL_FILE = "german.data"

# Column names follow the UCI documentation (german.doc), in file order
COLUMNS = ["status", "duration", "credit_history", "purpose", "amount", "savings", "employment",
           "installment_rate", "personal_status_sex", "other_debtors", "residence_since", "property",
           "age", "other_installment_plans", "housing", "existing_credits", "job", "num_dependents",
           "telephone", "foreign_worker", "target"]

def try_load(source, sep):
    try:
        d = pd.read_csv(source, sep=sep, header=None, names=COLUMNS, engine="python")
        return d if d.shape == (1000, 21) else None
    except Exception as e:
        print(f"  could not load {source}: {type(e).__name__}")
        return None

df = None
for source, sep in [(UCI_URL, r"\s+"), (LOCAL_FILE, None), (MIRROR_URL, ",")]:
    if source == LOCAL_FILE and not os.path.exists(LOCAL_FILE):
        continue
    print("trying", source)
    df = try_load(source, sep)
    if df is not None:
        print("  loaded from", source); record("data_source", source); break
assert df is not None, "Dataset could not be loaded from any source."
assert df.target.value_counts().to_dict() == {1: 700, 2: 300}, "Class counts differ from the UCI file (700 good / 300 bad)."

CODEBOOK = {   # readable labels for the categorical codes (UCI documentation)
 "status": {"A11": "checking < 0 DM", "A12": "checking 0-200 DM", "A13": "checking >= 200 DM / salary acct", "A14": "no checking account"},
 "credit_history": {"A30": "no credits / all paid", "A31": "all credits at this bank paid", "A32": "existing credits paid duly", "A33": "delay in past", "A34": "critical / other credits"},
 "purpose": {"A40": "new car", "A41": "used car", "A42": "furniture", "A43": "radio/TV", "A44": "appliances", "A45": "repairs", "A46": "education", "A47": "vacation", "A48": "retraining", "A49": "business", "A410": "others"},
 "savings": {"A61": "< 100 DM", "A62": "100-500 DM", "A63": "500-1000 DM", "A64": ">= 1000 DM", "A65": "unknown / none"},
 "employment": {"A71": "unemployed", "A72": "< 1 yr", "A73": "1-4 yrs", "A74": "4-7 yrs", "A75": ">= 7 yrs"},
 "personal_status_sex": {"A91": "male divorced/separated", "A92": "female div/sep/married", "A93": "male single", "A94": "male married/widowed", "A95": "female single"},
 "other_debtors": {"A101": "none", "A102": "co-applicant", "A103": "guarantor"},
 "property": {"A121": "real estate", "A122": "building society / life insurance", "A123": "car or other", "A124": "unknown / none"},
 "other_installment_plans": {"A141": "bank", "A142": "stores", "A143": "none"},
 "housing": {"A151": "rent", "A152": "own", "A153": "for free"},
 "job": {"A171": "unemployed / unskilled non-resident", "A172": "unskilled resident", "A173": "skilled", "A174": "management / self-employed"},
 "telephone": {"A191": "none", "A192": "yes"},
 "foreign_worker": {"A201": "yes", "A202": "no"},
}
def label(col):
    """'status_A14' -> 'status = no checking account (A14)'"""
    if "_A" in col:
        var, code = col.rsplit("_", 1)
        if var in CODEBOOK and code in CODEBOOK[var]:
            return f"{var} = {CODEBOOK[var][code]} ({code})"
    return col

# ------------------------------------------------------------------------------
# CELL 3 · Data inspection
# ------------------------------------------------------------------------------
show("Dataset shape", df.shape)
show("Columns", pd.DataFrame({"column": df.columns, "dtype": df.dtypes.values}))
missing_before = df.isna().sum().rename("missing_before")
show("Missing values BEFORE preprocessing", missing_before.to_frame())
print("placeholder tokens ('?', '', 'NA') present:", int(df.astype(str).isin(["?", "", " ", "NA"]).sum().sum()))
show("Target distribution (1 = good credit, 2 = bad credit)", df.target.value_counts().to_frame("count").assign(share=lambda t: t["count"] / len(df)))
show("Descriptive statistics – numeric attributes", df.describe().T)
NUMERIC_RAW = ["duration", "amount", "age", "installment_rate", "residence_since", "existing_credits", "num_dependents"]
CATEGORICAL = ["status", "credit_history", "purpose", "savings", "employment", "personal_status_sex", "other_debtors",
               "property", "other_installment_plans", "housing", "job", "telephone", "foreign_worker"]
show("Categorical value counts", pd.concat({c: df[c].value_counts() for c in CATEGORICAL}).to_frame("count"))
dataset_summary = pd.DataFrame({"item": ["rows", "attributes", "numeric attributes", "categorical attributes", "missing cells", "good (1)", "bad (2)"],
                                "value": [len(df), df.shape[1] - 1, len(NUMERIC_RAW), len(CATEGORICAL), int(df.isna().sum().sum()),
                                          int((df.target == 1).sum()), int((df.target == 2).sum())]})
dataset_summary.to_csv(f"{TAB_DIR}/01_dataset_summary.csv", index=False)
record("n_rows", len(df)); record("n_missing_cells", int(df.isna().sum().sum()))

fig, ax = plt.subplots(figsize=(4.5, 3.5))
df.target.map({1: "good (approve)", 2: "bad (reject)"}).value_counts().plot.bar(ax=ax, color=["tab:green", "tab:red"])
ax.set_title("Target distribution"); ax.set_ylabel("count"); plt.xticks(rotation=0); plt.tight_layout()
plt.savefig(f"{FIG_DIR}/fig01_target_distribution.png", dpi=150); plt.show()

# ------------------------------------------------------------------------------
# CELL 4 · Preprocessing
#   target encoding, missing values, derived group variables, outlier handling,
#   categorical encoding (one-hot), scaling is fitted on TRAIN ONLY in cell 5
# ------------------------------------------------------------------------------
TARGET_MAP = {1: 1, 2: 0}                                   # 1 = approve (good credit), 0 = reject (bad credit)
df["approve"] = df["target"].map(TARGET_MAP).astype(int)
print("Target encoding used:", {"UCI class 1 (good)": "approve = 1", "UCI class 2 (bad)": "approve = 0"})
show("Encoded target distribution", df.approve.value_counts().to_frame("count").assign(share=lambda t: t["count"] / len(df)))

# derived variables used ONLY to split the data in Parts 2.1 / 2.4 (the models use the original attributes)
df["sex"] = np.where(df.personal_status_sex.isin(["A92", "A95"]), "female", "male")
df["age_group"] = np.where(df.age <= 25, "young (<=25)", "older (>25)")
show("Approval rate by derived group variables",
     pd.concat([df.groupby("sex").approve.agg(["mean", "count"]), df.groupby("age_group").approve.agg(["mean", "count"])]))

# missing values – none exist, so no imputation is needed; recorded for the report
missing_after = df.isna().sum().rename("missing_after")
missing_table = pd.concat([missing_before, missing_after], axis=1).fillna(0).astype(int)
missing_table.to_csv(f"{TAB_DIR}/02_missing_values.csv")
show("Missing values BEFORE vs AFTER preprocessing", missing_table)
record("missing_handling", "no missing values in the UCI file; no imputation required")

# outlier analysis (IQR rule) and handling
outlier_rows = []
for c in ["duration", "amount", "age"]:
    q1, q3 = df[c].quantile([0.25, 0.75]); iqr = q3 - q1
    n_out = int(((df[c] < q1 - 1.5 * iqr) | (df[c] > q3 + 1.5 * iqr)).sum())
    outlier_rows.append([c, df[c].min(), df[c].median(), df[c].max(), round(df[c].skew(), 3), n_out])
df["log_amount"] = np.log(df["amount"])                     # log transform tames the right tail of 'amount'
q1, q3 = df.log_amount.quantile([0.25, 0.75]); iqr = q3 - q1
outlier_rows.append(["log_amount", round(df.log_amount.min(), 3), round(df.log_amount.median(), 3), round(df.log_amount.max(), 3),
                     round(df.log_amount.skew(), 3), int(((df.log_amount < q1 - 1.5 * iqr) | (df.log_amount > q3 + 1.5 * iqr)).sum())])
outlier_table = pd.DataFrame(outlier_rows, columns=["variable", "min", "median", "max", "skew", "IQR_outliers"])
outlier_table.to_csv(f"{TAB_DIR}/03_outlier_analysis.csv", index=False)
show("Outlier analysis (amount is log-transformed; duration and age kept – genuine values, models robust after scaling)", outlier_table)
record("outlier_handling", "log-transform of amount; duration/age retained; no rows removed")

# categorical encoding: one-hot with first level dropped (reference category) – keeps the design matrix full rank
NUMERIC = ["duration", "log_amount", "age", "installment_rate", "residence_since", "existing_credits", "num_dependents"]
X = pd.get_dummies(df[NUMERIC + CATEGORICAL], columns=CATEGORICAL, drop_first=True, dtype=float)
y = df["approve"]
reference_levels = {c: sorted(df[c].unique())[0] for c in CATEGORICAL}
print("\nNumerical columns  :", NUMERIC)
print("Categorical columns:", CATEGORICAL)
print("Reference (dropped) level of each categorical:", reference_levels)
show(f"Final feature list ({X.shape[1]} model columns)", pd.DataFrame({"feature": X.columns, "meaning": [label(c) for c in X.columns]}))
pd.DataFrame({"feature": X.columns}).to_csv(f"{TAB_DIR}/04_final_feature_list.csv", index=False)
record("n_model_features", X.shape[1])

# ------------------------------------------------------------------------------
# CELL 5 · Train/test split (70/30, random_state = 244) + scaling fitted on TRAIN only
#   No stratification: the assignment specifies only test_size and random_state, so the plain split is used.
# ------------------------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_SIZE, random_state=RANDOM_STATE)
scaler = StandardScaler().fit(X_train[NUMERIC])             # fitted on training rows only → no leakage
X_train_s, X_test_s = X_train.copy(), X_test.copy()
X_train_s[NUMERIC] = scaler.transform(X_train[NUMERIC]); X_test_s[NUMERIC] = scaler.transform(X_test[NUMERIC])
Xtr = sm.add_constant(X_train_s); Xte = sm.add_constant(X_test_s, has_constant="add")

split_table = pd.DataFrame({"rows": [len(Xtr), len(Xte)], "approve=1": [int(y_train.sum()), int(y_test.sum())],
                            "approve=0": [int((y_train == 0).sum()), int((y_test == 0).sum())],
                            "approval_rate": [y_train.mean(), y_test.mean()]}, index=["train", "test"]).round(4)
split_table.to_csv(f"{TAB_DIR}/05_train_test_split.csv")
show("Train / test split", split_table)
show("Scaler statistics (train only)", pd.DataFrame({"mean": scaler.mean_, "sd": scaler.scale_}, index=NUMERIC).round(4))
record("n_train", len(Xtr)); record("n_test", len(Xte)); record("train_approval_rate", y_train.mean()); record("test_approval_rate", y_test.mean())

# ------------------------------------------------------------------------------
# CELL 6 · Model A – OLS (linear probability model) on approve ∈ {0,1}
# ------------------------------------------------------------------------------
ols = sm.OLS(y_train, Xtr).fit()
pred_A = ols.predict(Xte)                                   # raw OLS predictions on the test set (kept as-is)
ols_table = pd.DataFrame({"coef": ols.params, "std_err": ols.bse, "t": ols.tvalues, "p_value": ols.pvalues,
                          "ci_low": ols.conf_int()[0], "ci_high": ols.conf_int()[1]}).round(4)
ols_table.insert(0, "meaning", [label(i) for i in ols_table.index])
ols_table.to_csv(f"{TAB_DIR}/06_ols_coefficients.csv")
show("Model A – OLS coefficient table (sorted by p-value)", ols_table.sort_values("p_value"))
print(ols.summary().tables[0])
median_A = float(np.median(pred_A))
record("ols_r2", ols.rsquared); record("ols_adj_r2", ols.rsquared_adj); record("ols_f_pvalue", ols.f_pvalue)
record("ols_test_rmse", float(np.sqrt(mean_squared_error(y_test, pred_A))))
record("ols_pred_median", median_A); record("ols_pred_min", float(pred_A.min())); record("ols_pred_max", float(pred_A.max()))
record("ols_pred_outside_0_1", int(((pred_A < 0) | (pred_A > 1)).sum()))
print(f"\nOLS test predictions: median = {median_A:.4f}, min = {pred_A.min():.4f}, max = {pred_A.max():.4f}, "
      f"{SUMMARY['ols_pred_outside_0_1']} of {len(pred_A)} outside [0,1]; R² = {ols.rsquared:.4f}")

# ------------------------------------------------------------------------------
# CELL 7 · Model B – Logistic regression (same split, same processed features)
# ------------------------------------------------------------------------------
logit = sm.Logit(y_train, Xtr).fit(disp=0, maxiter=500)
assert logit.mle_retvals["converged"], "logit did not converge"
prob_B = logit.predict(Xte)
class_B = (prob_B >= 0.5).astype(int)                       # threshold 0.5
logit_table = pd.DataFrame({"coef": logit.params, "std_err": logit.bse, "z": logit.tvalues, "p_value": logit.pvalues,
                            "ci_low": logit.conf_int()[0], "ci_high": logit.conf_int()[1], "odds_ratio": np.exp(logit.params)}).round(4)
logit_table.insert(0, "meaning", [label(i) for i in logit_table.index])
logit_table.to_csv(f"{TAB_DIR}/07_logistic_coefficients.csv")
show("Model B – logistic coefficient table (sorted by p-value)", logit_table.sort_values("p_value"))

cm_B = confusion_matrix(y_test, class_B, labels=[0, 1])
performance = pd.DataFrame({
    "metric": ["accuracy", "precision", "recall (TPR)", "F1", "ROC-AUC", "FPR", "pseudo R2 (McFadden)", "AIC"],
    "Model B (logit, t=0.5)": [accuracy_score(y_test, class_B), precision_score(y_test, class_B), recall_score(y_test, class_B),
                               f1_score(y_test, class_B), roc_auc_score(y_test, prob_B), cm_B[0, 1] / cm_B[0].sum(), logit.prsquared, logit.aic]}).round(4)
performance.to_csv(f"{TAB_DIR}/08_model_performance.csv", index=False)
show("Model B performance on the test set", performance)
show("Model B confusion matrix vs actual (rows = actual 0/1, cols = predicted 0/1)", pd.DataFrame(cm_B, index=["actual 0", "actual 1"], columns=["pred 0", "pred 1"]))
for m, v in zip(performance.metric, performance.iloc[:, 1]):
    record(f"logit_{m}", v)
record("logit_confusion_TN_FP_FN_TP", cm_B.ravel().tolist())

# ------------------------------------------------------------------------------
# CELL 8 · Part 1 – OLS median rule vs logistic 0.5 rule
# ------------------------------------------------------------------------------
class_A = (pred_A >= median_A).astype(int)                  # 1 if OLS prediction >= median
agreement = pd.crosstab(class_A.rename("OLS median rule"), class_B.rename("Logit 0.5 rule")).reindex(index=[0, 1], columns=[0, 1], fill_value=0)
n_match = int((class_A == class_B).sum()); n_mismatch = int((class_A != class_B).sum())
mismatch_stats = pd.DataFrame({"statistic": ["median OLS prediction", "flagged 1 by OLS rule", "flagged 1 by logit rule", "matching cases",
                                             "mismatched cases", "% matched", "% mismatched", "OLS=0 & logit=1", "OLS=1 & logit=0",
                                             "accuracy of OLS median rule vs truth", "accuracy of logit 0.5 rule vs truth",
                                             "actual approval rate among mismatched", "corr(OLS prediction, logit probability)"],
                               "value": [median_A, int(class_A.sum()), int(class_B.sum()), n_match, n_mismatch,
                                         100 * n_match / len(class_A), 100 * n_mismatch / len(class_A),
                                         int(((class_A == 0) & (class_B == 1)).sum()), int(((class_A == 1) & (class_B == 0)).sum()),
                                         accuracy_score(y_test, class_A), accuracy_score(y_test, class_B),
                                         float(y_test[class_A != class_B].mean()), float(np.corrcoef(pred_A, prob_B)[0, 1])]}).round(4)
agreement.to_csv(f"{TAB_DIR}/09_ols_vs_logit_confusion.csv"); mismatch_stats.to_csv(f"{TAB_DIR}/10_ols_vs_logit_mismatch.csv", index=False)
show("Part 1 – confusion / agreement matrix (rows = OLS median rule, cols = logit 0.5 rule)", agreement)
show("Part 1 – mismatch statistics", mismatch_stats)
mm = class_A != class_B
print(f"Mismatched rows: OLS predictions range {pred_A[mm].min():.4f}–{pred_A[mm].max():.4f}; logit probabilities range {prob_B[mm].min():.4f}–{prob_B[mm].max():.4f}")
for s, v in zip(mismatch_stats.statistic, mismatch_stats.value):
    record(f"part1_{s}", v)

fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))
sns.heatmap(agreement, annot=True, fmt="d", cmap="Blues", cbar=False, ax=ax[0]); ax[0].set_title(f"OLS median rule vs logit 0.5 rule\n(mismatch = {n_mismatch}, {100*n_mismatch/len(class_A):.1f}%)")
ax[1].scatter(pred_A, prob_B, c=np.where(mm, "tab:red", "tab:blue"), s=14, alpha=.7); ax[1].axvline(median_A, ls="--", c="k"); ax[1].axhline(0.5, ls="--", c="k")
ax[1].set_xlabel("OLS prediction"); ax[1].set_ylabel("logit probability"); ax[1].set_title("red = classified differently")
ax[2].hist(pred_A, bins=25, alpha=.6, label="OLS prediction"); ax[2].hist(prob_B, bins=25, alpha=.6, label="logit probability")
ax[2].axvline(median_A, ls="--", c="tab:blue"); ax[2].axvline(0.5, ls="--", c="tab:orange"); ax[2].legend(); ax[2].set_title("score distributions and thresholds")
plt.tight_layout(); plt.savefig(f"{FIG_DIR}/fig02_model_comparison.png", dpi=150); plt.show()

# ------------------------------------------------------------------------------
# CELL 9 · Part 2.1 – Simpson's paradox / subgroup analysis
#   Split variable: sex. Fits on the full data (1000 rows) and on each sex with a compact,
#   identical specification (subgroups are too small for all 48 dummies). Every predictor is
#   compared; the predictor with the most significant female–male difference is reported.
# ------------------------------------------------------------------------------
SPLIT_VAR, GROUP_A, GROUP_B = "sex", "female", "male"
S_NUM = ["duration", "log_amount", "age", "installment_rate"]; S_CAT = ["status", "credit_history", "savings", "employment"]
def compact_design(sub):
    Xs = pd.get_dummies(sub[S_NUM + S_CAT], columns=S_CAT, drop_first=True, dtype=float)
    idx = [NUMERIC.index(c) for c in S_NUM]
    Xs[S_NUM] = (Xs[S_NUM] - scaler.mean_[idx]) / scaler.scale_[idx]        # same scale as Model B
    return sm.add_constant(Xs)
sub_fits = {"Full dataset": sm.Logit(df.approve, compact_design(df)).fit(disp=0, maxiter=500)}
for g in (GROUP_A, GROUP_B):
    m = df[SPLIT_VAR] == g
    sub_fits[f"Subgroup {g}"] = sm.Logit(df.approve[m], compact_design(df[m])).fit(disp=0, maxiter=500)

rows = []
for p in sub_fits["Full dataset"].params.index.drop("const"):
    fA, fB = sub_fits[f"Subgroup {GROUP_A}"], sub_fits[f"Subgroup {GROUP_B}"]
    z = (fA.params[p] - fB.params[p]) / np.sqrt(fA.bse[p] ** 2 + fB.bse[p] ** 2)
    rows.append({"predictor": p, "full_coef": sub_fits["Full dataset"].params[p], "full_p": sub_fits["Full dataset"].pvalues[p],
                 f"{GROUP_A}_coef": fA.params[p], f"{GROUP_A}_p": fA.pvalues[p], f"{GROUP_B}_coef": fB.params[p], f"{GROUP_B}_p": fB.pvalues[p],
                 "z_difference": z, "p_difference": 2 * (1 - stats.norm.cdf(abs(z))),
                 "sign_reversal": np.sign(fA.params[p]) != np.sign(fB.params[p])})
all_predictors = pd.DataFrame(rows).sort_values("p_difference").round(4)
all_predictors.to_csv(f"{TAB_DIR}/11a_subgroup_all_predictors.csv", index=False)
show("Part 2.1 – all predictors: full vs female vs male (sorted by significance of the female–male difference)", all_predictors)

KEY = all_predictors.iloc[0]["predictor"]                    # data-chosen key predictor
subgroup_results = pd.DataFrame([{"Model": name, "Group": name.replace("Subgroup ", "") if "Subgroup" in name else "all", "Predictor": KEY,
                                  "Coefficient": f.params[KEY], "Std Error": f.bse[KEY], "P-value": f.pvalues[KEY],
                                  "CI Lower": f.conf_int().loc[KEY, 0], "CI Upper": f.conf_int().loc[KEY, 1], "N": int(f.nobs)}
                                 for name, f in sub_fits.items()]).round(4)
subgroup_results.to_csv(f"{TAB_DIR}/11_subgroup_regression_results.csv", index=False); subgroup_results.to_csv(f"{OUT_DIR}/subgroup_results.csv", index=False)
show(f"Part 2.1 – key predictor '{KEY}' across the three models", subgroup_results)
key_row = all_predictors.set_index("predictor").loc[KEY]
sd_key = scaler.scale_[NUMERIC.index(KEY)] if KEY in NUMERIC else np.nan
reversal = bool(key_row.sign_reversal)
sig_A, sig_B = key_row[f"{GROUP_A}_p"] < 0.05, key_row[f"{GROUP_B}_p"] < 0.05
print("\nMACHINE-GENERATED INTERPRETATION SUMMARY")
print(f"  Key predictor: {KEY}" + (f" (coefficient per 1 SD = {sd_key:.2f} units)" if KEY in NUMERIC else ""))
print(f"  Full data   : {key_row.full_coef:+.4f} (p = {key_row.full_p:.4f})")
print(f"  {GROUP_A:<12}: {key_row[f'{GROUP_A}_coef']:+.4f} (p = {key_row[f'{GROUP_A}_p']:.4f}) {'significant' if sig_A else 'not significant'} at 5%")
print(f"  {GROUP_B:<12}: {key_row[f'{GROUP_B}_coef']:+.4f} (p = {key_row[f'{GROUP_B}_p']:.4f}) {'significant' if sig_B else 'not significant'} at 5%")
print(f"  Difference  : z = {key_row.z_difference:.3f}, p = {key_row.p_difference:.4f}")
if reversal and sig_A and sig_B:
    verdict = "GENUINE SIGN REVERSAL: both subgroup coefficients are significant and have opposite signs (Simpson-type reversal)."
elif reversal:
    verdict = ("Sign differs between subgroups and the difference is significant, but the coefficient in one subgroup is not itself "
               "significantly different from zero → report as 'significantly different effect with an observed sign flip', not a proven reversal."
               if key_row.p_difference < 0.05 else "Signs differ but the difference is not significant → no reversal can be claimed.")
else:
    verdict = ("No sign reversal; the effect is significantly stronger in one subgroup (assignment fallback applies)."
               if key_row.p_difference < 0.05 else "No reversal and no significant difference for this predictor.")
print("  Verdict     :", verdict)
record("simpson_split_variable", SPLIT_VAR); record("simpson_key_predictor", KEY); record("simpson_verdict", verdict)
for name, f in sub_fits.items():
    record(f"simpson_{name}_coef", f.params[KEY]); record(f"simpson_{name}_p", f.pvalues[KEY]); record(f"simpson_{name}_ci", f.conf_int().loc[KEY].round(4).tolist())
record("simpson_z_difference", key_row.z_difference); record("simpson_p_difference", key_row.p_difference)

df["age_band"] = pd.cut(df.age, [0, 25, 35, 45, 100], labels=["<=25", "26-35", "36-45", "46+"])
rate_by_band = df.groupby([SPLIT_VAR, "age_band"], observed=True).approve.agg(rate="mean", n="count").round(3).unstack("age_band")
rate_by_band.to_csv(f"{TAB_DIR}/11b_approval_rate_by_sex_ageband.csv")
show("Part 2.1 – raw approval rate by sex × age band (descriptive support)", rate_by_band)

fig, ax = plt.subplots(1, 2, figsize=(12, 4.5))
for i, (name, f) in enumerate(sub_fits.items()):
    lo, hi = f.conf_int().loc[KEY]
    ax[0].errorbar(f.params[KEY], i, xerr=[[f.params[KEY] - lo], [hi - f.params[KEY]]], fmt="o", capsize=4)
ax[0].axvline(0, c="k"); ax[0].set_yticks(range(3)); ax[0].set_yticklabels([f"{n} (n={int(f.nobs)})" for n, f in sub_fits.items()])
ax[0].set_xlabel(f"logit coefficient of {KEY} with 95% CI"); ax[0].set_title(f"Effect of {KEY} on approval: full vs by {SPLIT_VAR}")
for g in (GROUP_A, GROUP_B):
    ax[1].plot(rate_by_band.loc[g, "rate"].index.astype(str), rate_by_band.loc[g, "rate"].values, marker="o", label=g)
ax[1].set_xlabel("age band"); ax[1].set_ylabel("approval rate"); ax[1].legend(); ax[1].set_title("Approval rate by age band within sex")
plt.tight_layout(); plt.savefig(f"{FIG_DIR}/fig03_subgroup_analysis.png", dpi=150); plt.show()

# ------------------------------------------------------------------------------
# CELL 10 · Part 2.2 – Omitted-variable bias (confounder: checking-account 'status')
#   Full model = Model B (training data). Reduced model = same without the 3 status dummies.
# ------------------------------------------------------------------------------
CONFOUNDER = "status"
kept = [c for c in Xtr.columns if not c.startswith(CONFOUNDER + "_")]
logit_reduced = sm.Logit(y_train, Xtr[kept]).fit(disp=0, maxiter=500)
ovb = pd.DataFrame({"full_coef": logit.params[kept], "reduced_coef": logit_reduced.params[kept]}).drop("const")
ovb["abs_change"] = (ovb.reduced_coef - ovb.full_coef).abs()
ovb["pct_change"] = np.where(ovb.full_coef.abs() > 1e-12, ovb.abs_change / ovb.full_coef.abs() * 100, np.nan)   # safe for zero coefficient
ovb["full_p"] = logit.pvalues[kept].drop("const"); ovb.insert(0, "meaning", [label(i) for i in ovb.index])
ovb_full_table = ovb.round(4); ovb_full_table.to_csv(f"{TAB_DIR}/12a_omitted_variable_all_predictors.csv")
FOCUS = ["savings_A65", "credit_history_A34", "duration", "installment_rate", "log_amount", "age", "other_installment_plans_A143", "housing_A152"]
ovb_focus = ovb.loc[FOCUS].round(4)
ovb_focus.to_csv(f"{TAB_DIR}/12_omitted_variable_comparison.csv"); ovb_focus.to_csv(f"{OUT_DIR}/omitted_variable_results.csv")
show(f"Part 2.2 – before/after dropping '{CONFOUNDER}' (focus predictors)", ovb_focus)
show("Part 2.2 – all predictors sorted by % change", ovb_full_table.sort_values("pct_change", ascending=False))
largest_abs = ovb.abs_change.idxmax(); largest_pct = ovb["pct_change"].idxmax()
largest_pct_sig = ovb[ovb.full_p < 0.05]["pct_change"].idxmax(); largest_pct_focus = ovb.loc[FOCUS]["pct_change"].idxmax()
lr_stat = 2 * (logit.llf - logit_reduced.llf); lr_p = 1 - stats.chi2.cdf(lr_stat, df=len(Xtr.columns) - len(kept))
print(f"Largest absolute shift : {largest_abs} ({ovb.loc[largest_abs,'abs_change']:.4f}; {ovb.loc[largest_abs,'full_coef']:.4f} → {ovb.loc[largest_abs,'reduced_coef']:.4f})")
print(f"Largest % shift (all)  : {largest_pct} ({ovb.loc[largest_pct,'pct_change']:.1f}%; full coef {ovb.loc[largest_pct,'full_coef']:.4f} – near-zero base inflates %)")
print(f"Largest % shift (focus): {largest_pct_focus} ({ovb.loc[largest_pct_focus,'pct_change']:.1f}%)")
print(f"Largest % shift (p<.05): {largest_pct_sig} ({ovb.loc[largest_pct_sig,'pct_change']:.1f}%)")
print(f"Predictors moving >10%: {int((ovb['pct_change'] > 10).sum())} of {len(ovb)}; median % change {ovb['pct_change'].median():.2f}%")
print(f"LR test for dropped dummies: chi2 = {lr_stat:.2f}, p = {lr_p:.2e}; AIC {logit.aic:.1f} → {logit_reduced.aic:.1f}")
red_prob = logit_reduced.predict(Xte[kept])
print(f"Test accuracy {accuracy_score(y_test, class_B):.4f} → {accuracy_score(y_test, (red_prob>=0.5).astype(int)):.4f}; AUC {roc_auc_score(y_test, prob_B):.4f} → {roc_auc_score(y_test, red_prob):.4f}")
record("ovb_dropped_variable", CONFOUNDER); record("ovb_largest_abs_shift", largest_abs); record("ovb_largest_pct_shift_all", largest_pct)
record("ovb_largest_pct_shift_focus", largest_pct_focus); record("ovb_largest_pct_shift_significant", largest_pct_sig)
record("ovb_n_predictors_over_10pct", int((ovb["pct_change"] > 10).sum())); record("ovb_LR_chi2", lr_stat); record("ovb_LR_p", lr_p)
record("ovb_reduced_test_accuracy", accuracy_score(y_test, (red_prob >= 0.5).astype(int))); record("ovb_reduced_test_auc", roc_auc_score(y_test, red_prob))
for p in FOCUS:
    record(f"ovb_{p}_full", ovb.loc[p, "full_coef"]); record(f"ovb_{p}_reduced", ovb.loc[p, "reduced_coef"]); record(f"ovb_{p}_pct_change", ovb.loc[p, "pct_change"])

fig, ax = plt.subplots(figsize=(10, 4.5)); xx = np.arange(len(FOCUS)); w = 0.38
ax.bar(xx - w / 2, ovb_focus.full_coef, w, label="full model"); ax.bar(xx + w / 2, ovb_focus.reduced_coef, w, label=f"reduced model (no {CONFOUNDER})")
for i, p in enumerate(FOCUS):
    ax.text(i, max(ovb_focus.full_coef[p], ovb_focus.reduced_coef[p], 0) + 0.03, f"{ovb_focus['pct_change'][p]:.0f}%", ha="center", fontsize=8)
ax.axhline(0, c="k", lw=.8); ax.set_xticks(xx); ax.set_xticklabels(FOCUS, rotation=30, ha="right"); ax.legend(); ax.set_title("Omitted-variable test: coefficient shifts (labels = % change)")
plt.tight_layout(); plt.savefig(f"{FIG_DIR}/fig04_omitted_variable.png", dpi=150); plt.show()

# ------------------------------------------------------------------------------
# CELL 11 · Part 2.3 – Adversarial subset (test set only)
# ------------------------------------------------------------------------------
test_df = df.loc[X_test.index].copy()
test_df["prob_approve"] = prob_B.values; test_df["actual"] = y_test.values; test_df["predicted"] = class_B.values
fp_mask = (test_df.prob_approve >= 0.75) & (test_df.actual == 0)
fn_mask = (test_df.prob_approve <= 0.25) & (test_df.actual == 1)
confident_wrong = test_df[fp_mask | fn_mask].copy()
confident_wrong["error_type"] = np.where(confident_wrong.actual == 0, "false positive", "false negative")
n_conf = len(confident_wrong)
if n_conf > 50:
    adversarial = confident_wrong.sample(n=50, random_state=SAMPLING_SEED)
    sampling_note = f"{n_conf} confident errors > 50 → 50 sampled with random_state = {SAMPLING_SEED}"
else:
    adversarial = confident_wrong.copy()
    sampling_note = f"{n_conf} confident errors (< 50) → all {n_conf} rows used"
print(sampling_note)
adv_summary = pd.DataFrame({"statistic": ["high-confidence wrong predictions", "% of test set", "false positives (P>=0.75, actual 0)",
                                          "false negatives (P<=0.25, actual 1)", "rows in adversarial subset", "mean P among FP", "mean P among FN",
                                          "FP as % of all actually-bad test rows", "% of test rows with P>=0.75", "% of test rows with P<=0.25"],
                            "value": [n_conf, 100 * n_conf / len(test_df), int(fp_mask.sum()), int(fn_mask.sum()), len(adversarial),
                                      adversarial.loc[adversarial.actual == 0, "prob_approve"].mean(), adversarial.loc[adversarial.actual == 1, "prob_approve"].mean(),
                                      100 * fp_mask.sum() / (test_df.actual == 0).sum(), 100 * (test_df.prob_approve >= 0.75).mean(), 100 * (test_df.prob_approve <= 0.25).mean()]}).round(4)
adv_summary.to_csv(f"{TAB_DIR}/13_adversarial_subset_summary.csv", index=False)
show("Part 2.3 – adversarial subset summary", adv_summary)
adv_cols = ["prob_approve", "actual", "predicted", "error_type", "age", "sex", "age_group", "status", "credit_history", "savings", "purpose",
            "installment_rate", "employment", "duration", "amount", "housing", "job", "foreign_worker"]
adversarial_out = adversarial[adv_cols].sort_values("prob_approve", ascending=False).round(4)
adversarial_out.to_csv(f"{OUT_DIR}/adversarial_subset.csv"); adversarial_out.to_csv(f"{TAB_DIR}/14_adversarial_subset_rows.csv")
show("Part 2.3 – adversarial subset rows", adversarial_out)
for s, v in zip(adv_summary.statistic, adv_summary.value):
    record(f"adv_{s}", v)

def demographics(d):
    return pd.Series({"n": len(d), "age mean": d.age.mean(), "age median": d.age.median(), "age <=25 %": 100 * (d.age <= 25).mean(),
                      "age 26-35 %": 100 * d.age.between(26, 35).mean(), "age 36-45 %": 100 * d.age.between(36, 45).mean(), "age 46+ %": 100 * (d.age > 45).mean(),
                      "female %": 100 * (d.sex == "female").mean(), "male %": 100 * (d.sex == "male").mean(),
                      "foreign worker %": 100 * (d.foreign_worker == "A201").mean(), "own housing %": 100 * (d.housing == "A152").mean(),
                      "skilled job (A173) %": 100 * (d.job == "A173").mean(), "installment_rate mean": d.installment_rate.mean(),
                      "installment_rate = 4 %": 100 * (d.installment_rate == 4).mean(), "duration mean": d.duration.mean(), "amount median": d.amount.median()})
demo_table = pd.concat({"adversarial subset": demographics(adversarial), "full test set": demographics(test_df)}, axis=1).round(2)
demo_table["difference"] = (demo_table.iloc[:, 0] - demo_table.iloc[:, 1]).round(2)
demo_table.to_csv(f"{TAB_DIR}/15_adversarial_vs_test_demographics.csv")
show("Part 2.3 – adversarial subset vs full test set demographics", demo_table)

# automatic over-representation search: every binary indicator, share in subset vs test set, Fisher exact test
indicators = pd.DataFrame(index=test_df.index)
for c in X_test.columns:
    if c not in NUMERIC:
        indicators[label(c)] = X_test[c]
indicators["installment_rate = 4"] = (test_df.installment_rate == 4).astype(int)
indicators["installment_rate >= 3"] = (test_df.installment_rate >= 3).astype(int)
indicators["age <= 25"] = (test_df.age <= 25).astype(int); indicators["female"] = (test_df.sex == "female").astype(int)
indicators["duration > 24"] = (test_df.duration > 24).astype(int); indicators["amount > test median"] = (test_df.amount > test_df.amount.median()).astype(int)
indicators["savings < 100 DM (A61)"] = (test_df.savings == "A61").astype(int); indicators["checking < 0 DM (A11)"] = (test_df.status == "A11").astype(int)
indicators["credit_history A30/A31"] = test_df.credit_history.isin(["A30", "A31"]).astype(int)
indicators["employment unemployed or < 1 yr (A71/A72)"] = test_df.employment.isin(["A71", "A72"]).astype(int)
indicators["purpose education/retraining/other (A46/A48/A410)"] = test_df.purpose.isin(["A46", "A48", "A410"]).astype(int)
in_adv = indicators.index.isin(adversarial.index)
rows = []
for c in indicators.columns:
    a, b = indicators.loc[in_adv, c], indicators.loc[~in_adv, c]
    p = stats.fisher_exact([[a.sum(), len(a) - a.sum()], [b.sum(), len(b) - b.sum()]])[1]
    rows.append({"characteristic": c, "adv_count": int(a.sum()), "adv_%": 100 * a.mean(), "test_%": 100 * indicators[c].mean(),
                 "difference_pp": 100 * (a.mean() - indicators[c].mean()), "fisher_p": p})
overrep = pd.DataFrame(rows).sort_values("fisher_p").round(4)
overrep.to_csv(f"{TAB_DIR}/16_adversarial_overrepresented_characteristics.csv", index=False)
show("Part 2.3 – over/under-represented characteristics (sorted by Fisher p)", overrep.head(20))
majority = overrep[(overrep["adv_%"] >= 50) & (overrep.fisher_p < 0.05)].sort_values("difference_pp", ascending=False)
strongest = majority.iloc[0] if len(majority) else overrep.iloc[0]
print("\nStrongest common characteristic (shared by a majority of the subset and significantly over-represented):")
print(strongest.to_string())
print("\nMost over-represented characteristic (largest gap, any share):")
print(overrep.sort_values("difference_pp", ascending=False).iloc[0].to_string())
record("adv_strongest_characteristic", strongest.characteristic); record("adv_strongest_adv_%", strongest["adv_%"]); record("adv_strongest_test_%", strongest["test_%"]); record("adv_strongest_fisher_p", strongest.fisher_p)
record("adv_most_overrepresented", overrep.sort_values("difference_pp", ascending=False).iloc[0].characteristic)
# decision-boundary evidence from Model B relevant to the subset
ir_unit = logit.params["installment_rate"] / scaler.scale_[NUMERIC.index("installment_rate")]
print(f"\nModel B evidence: installment_rate coef = {logit.params['installment_rate']:.4f} per SD = {ir_unit:.4f} per unit (p = {logit.pvalues['installment_rate']:.4f}); "
      f"odds ratio rate 4 vs 1 = {np.exp(3*ir_unit):.3f}; status_A14 coef = {logit.params['status_A14']:.4f} (OR {np.exp(logit.params['status_A14']):.2f})")
for c in ["purpose_A410", "purpose_A46", "purpose_A48", "purpose_A45"]:
    if c in logit.params:
        print(f"  {c}: coef {logit.params[c]:.4f}, SE {logit.bse[c]:.4f}, training rows = {int(X_train[c].sum())}")
X_int = Xtr.copy(); X_int["installment_x_A14"] = X_int.installment_rate * X_int.status_A14
lg_int = sm.Logit(y_train, X_int).fit(disp=0, maxiter=500)
print(f"  interaction installment_rate × status_A14: coef {lg_int.params['installment_x_A14']:.4f}, p = {lg_int.pvalues['installment_x_A14']:.4f}")
record("adv_installment_rate_coef_per_unit", ir_unit); record("adv_interaction_p", lg_int.pvalues["installment_x_A14"])

fig, ax = plt.subplots(1, 2, figsize=(13, 4.5))
top = overrep.sort_values("difference_pp", key=abs, ascending=False).head(8); yy = np.arange(len(top)); w = .38
ax[0].barh(yy - w / 2, top["adv_%"], w, label=f"adversarial (n={len(adversarial)})"); ax[0].barh(yy + w / 2, top["test_%"], w, label=f"test set (n={len(test_df)})")
ax[0].set_yticks(yy); ax[0].set_yticklabels([c[:45] for c in top.characteristic], fontsize=8); ax[0].invert_yaxis(); ax[0].set_xlabel("%"); ax[0].legend(fontsize=8); ax[0].set_title("Most over/under-represented characteristics")
ax[1].hist(test_df.age, bins=15, alpha=.5, density=True, label="test set"); ax[1].hist(adversarial.age, bins=15, alpha=.5, density=True, label="adversarial"); ax[1].legend(); ax[1].set_xlabel("age"); ax[1].set_title("Age distribution")
plt.tight_layout(); plt.savefig(f"{FIG_DIR}/fig05_adversarial_subset.png", dpi=150); plt.show()

# ------------------------------------------------------------------------------
# CELL 12 · Part 2.4 – Fairness sweep (sensitive variable: age; A = young <=25, B = older >25)
# ------------------------------------------------------------------------------
SENSITIVE = "age_group"; FAIR_A = "young (<=25)"; FAIR_B = "older (>25)"
print(f"Sensitive variable: age. Group A = {FAIR_A} (age <= 25). Group B = {FAIR_B} (age > 25).")
show("Group sizes and actual approval rates in the test set", test_df.groupby(SENSITIVE).actual.agg(n="count", approval_rate="mean").round(4))
THRESHOLDS = np.round(np.arange(0.05, 1.0, 0.10), 2)
def safe_ratio(num, den):
    return np.nan if den == 0 else num / den            # undefined → NaN (documented), never a fake number

def fairness_sweep(d, col, A, B):
    out = []
    for t in THRESHOLDS:
        yhat = (d.prob_approve >= t).astype(int); r = {"threshold": t}
        for g, val in (("A", A), ("B", B)):
            m = d[col] == val
            r[f"pos_rate_{g}"] = yhat[m].mean()
            r[f"TPR_{g}"] = yhat[m & (d.actual == 1)].mean(); r[f"FPR_{g}"] = yhat[m & (d.actual == 0)].mean()
            r[f"PPV_{g}"] = d.actual[m & (yhat == 1)].mean() if (m & (yhat == 1)).sum() > 0 else np.nan
            r[f"accuracy_{g}"] = (yhat[m] == d.actual[m]).mean()
        r["DP"] = safe_ratio(r["pos_rate_A"], r["pos_rate_B"])
        r["TPR_ratio"] = safe_ratio(r["TPR_A"], r["TPR_B"]); r["FPR_ratio"] = safe_ratio(r["FPR_A"], r["FPR_B"])
        r["EO"] = np.nan if np.isnan(r["TPR_ratio"]) and np.isnan(r["FPR_ratio"]) else np.nanmin([r["TPR_ratio"], r["FPR_ratio"]])
        r["PP"] = np.nan if np.isnan(r["PPV_A"]) or np.isnan(r["PPV_B"]) else safe_ratio(r["PPV_A"], r["PPV_B"])
        r["overall_accuracy"] = (yhat == d.actual).mean()
        out.append(r)
    return pd.DataFrame(out)
fair = fairness_sweep(test_df, SENSITIVE, FAIR_A, FAIR_B)
fair_table = fair[["threshold", "DP", "EO", "PP"]].round(4)
group_stats = fair[["threshold", "pos_rate_A", "pos_rate_B", "TPR_A", "TPR_B", "FPR_A", "FPR_B", "PPV_A", "PPV_B", "accuracy_A", "accuracy_B", "overall_accuracy"]].round(4)
fair.round(4).to_csv(f"{OUT_DIR}/fairness_metrics.csv", index=False); fair_table.to_csv(f"{TAB_DIR}/17_fairness_DP_EO_PP.csv", index=False)
group_stats.to_csv(f"{TAB_DIR}/18_fairness_group_specific.csv", index=False)
show("Part 2.4 – DP / EO / PP at every threshold (NaN = undefined, zero denominator)", fair_table)
show("Part 2.4 – group-specific rates", group_stats)
n_undefined = int(fair[["DP", "EO", "PP"]].isna().sum().sum())
print(f"Undefined (NaN) metric cells: {n_undefined}")
record("fair_sensitive_variable", "age (A: <=25, B: >25)"); record("fair_undefined_cells", n_undefined)
for _, r in fair.iterrows():
    record(f"fair_DP_t{r.threshold:.2f}", r.DP); record(f"fair_EO_t{r.threshold:.2f}", r.EO); record(f"fair_PP_t{r.threshold:.2f}", r.PP); record(f"fair_acc_t{r.threshold:.2f}", r.overall_accuracy)

# supplementary sweep on sex (robustness) – saved but the age split is the primary analysis
fair_sex = fairness_sweep(test_df, "sex", "female", "male")
fair_sex.round(4).to_csv(f"{TAB_DIR}/18b_fairness_supplementary_sex.csv", index=False)
show("Part 2.4 – supplementary sweep with sex (A = female, B = male)", fair_sex[["threshold", "DP", "EO", "PP", "overall_accuracy"]].round(4))

# ------------------------------------------------------------------------------
# CELL 13 · Fairness trade-off chart (DP on x, EO on y, both 0–1.5, each point labelled)
# ------------------------------------------------------------------------------
plot_df = fair.dropna(subset=["DP", "EO"])
fig, ax = plt.subplots(figsize=(7, 6.5))
ax.plot(plot_df.DP, plot_df.EO, "-", color="lightgrey", zorder=1)
sc = ax.scatter(plot_df.DP, plot_df.EO, c=plot_df.threshold, cmap="viridis", s=80, edgecolor="k", zorder=2, label="threshold t")
offsets = [(8, 6), (-42, 10), (8, -14), (-44, -12), (8, 6), (8, -12), (8, 6), (-46, -12), (8, 6), (8, 6)]
for i, (_, r) in enumerate(plot_df.iterrows()):
    ax.annotate(f"t={r.threshold:.2f}", (r.DP, r.EO), textcoords="offset points", xytext=offsets[i % len(offsets)], fontsize=8,
                arrowprops=dict(arrowstyle="-", color="grey", lw=.5))
ax.plot(1, 1, "r*", ms=14, label="ideal (1, 1)"); ax.axvline(1, ls=":", c="k", lw=1); ax.axhline(1, ls=":", c="k", lw=1)
ax.set_xlim(0, 1.5); ax.set_ylim(0, 1.5); ax.grid(True, alpha=.4)
ax.set_xlabel(f"Demographic Parity  P(Ŷ=1 | {FAIR_A}) / P(Ŷ=1 | {FAIR_B})"); ax.set_ylabel("Equalized Odds  min(TPR_A/TPR_B, FPR_A/FPR_B)")
ax.set_title("Fairness trade-off across decision thresholds (test set)"); ax.legend(loc="lower right"); plt.colorbar(sc, label="threshold")
plt.tight_layout(); plt.savefig(f"{FIG_DIR}/fig06_fairness_tradeoff_DP_vs_EO.png", dpi=150); plt.show()

fig, ax = plt.subplots(figsize=(9, 4.5))
for m in ["DP", "EO", "PP", "overall_accuracy"]:
    ax.plot(fair.threshold, fair[m], marker="o", label=m)
ax.axhline(1, ls=":", c="k"); ax.set_xticks(THRESHOLDS); ax.set_xlabel("threshold"); ax.set_ylabel("value"); ax.legend(); ax.set_title("DP, EO, PP and accuracy vs threshold")
plt.tight_layout(); plt.savefig(f"{FIG_DIR}/fig07_metrics_vs_threshold.png", dpi=150); plt.show()

# ------------------------------------------------------------------------------
# CELL 14 · Trade-off calculations (assignment formula: cost = EO loss % / DP gain %)
# ------------------------------------------------------------------------------
def pct_change(a, b):
    return np.nan if (a == 0 or np.isnan(a) or np.isnan(b)) else 100 * (b - a) / abs(a)

def tradeoff(t_from, t_to):
    a = fair.set_index("threshold").loc[t_from]; b = fair.set_index("threshold").loc[t_to]
    dp_gain = pct_change(a.DP, b.DP); eo_change = pct_change(a.EO, b.EO); pp_change = pct_change(a.PP, b.PP)
    return {"from_t": t_from, "to_t": t_to, "DP_from": a.DP, "DP_to": b.DP, "DP_%change": dp_gain,
            "EO_from": a.EO, "EO_to": b.EO, "EO_%change": eo_change, "PP_from": a.PP, "PP_to": b.PP, "PP_%change": pp_change,
            "acc_from": a.overall_accuracy, "acc_to": b.overall_accuracy,
            "tradeoff_cost_EO_loss_per_DP_gain": (-eo_change / dp_gain) if (dp_gain not in (0, np.nan) and not np.isnan(dp_gain)) else np.nan,
            "tradeoff_cost_PP_loss_per_DP_gain": (-pp_change / dp_gain) if (dp_gain not in (0, np.nan) and not np.isnan(dp_gain)) else np.nan}
consecutive = pd.DataFrame([tradeoff(THRESHOLDS[i], THRESHOLDS[i + 1]) for i in range(len(THRESHOLDS) - 1)]).round(4)
consecutive.to_csv(f"{TAB_DIR}/19a_tradeoff_consecutive_thresholds.csv", index=False)
show("Trade-offs between consecutive thresholds", consecutive)
best = {"most accurate": float(fair.loc[fair.overall_accuracy.idxmax(), "threshold"]),
        "DP closest to 1": float(fair.loc[(fair.DP - 1).abs().idxmin(), "threshold"]),
        "EO closest to 1": float(fair.loc[(fair.EO - 1).abs().idxmin(), "threshold"]),
        "PP closest to 1": float(fair.loc[(fair.PP - 1).abs().idxmin(), "threshold"])}
print("Threshold achieving the best value of each criterion:", best)
pairs = sorted({(best["most accurate"], round(best["most accurate"] - 0.10, 2)), (best["PP closest to 1"], best["DP closest to 1"]),
                (best["most accurate"], best["PP closest to 1"]), (0.45, 0.35), (0.65, 0.35), (0.35, 0.25)})
selected = pd.DataFrame([tradeoff(a, b) for a, b in pairs if a in THRESHOLDS and b in THRESHOLDS and a != b]).round(4)
selected.to_csv(f"{TAB_DIR}/19_tradeoff_calculations.csv", index=False)
show("Selected trade-off calculations (evidence for the written report)", selected)
for _, r in selected.iterrows():
    print(f"t {r.from_t:.2f} → {r.to_t:.2f}: DP {r.DP_from:.3f}→{r.DP_to:.3f} ({r['DP_%change']:+.2f}%), EO {r.EO_from:.3f}→{r.EO_to:.3f} ({r['EO_%change']:+.2f}%), "
          f"PP {r.PP_from:.3f}→{r.PP_to:.3f} ({r['PP_%change']:+.2f}%), accuracy {r.acc_from:.3f}→{r.acc_to:.3f}, cost(EO/DP) = {r.tradeoff_cost_EO_loss_per_DP_gain:.3f}")
for k, v in best.items():
    record(f"fair_best_threshold_{k}", v)
for _, r in selected.iterrows():
    record(f"tradeoff_{r.from_t:.2f}_to_{r.to_t:.2f}_DP_pct", r["DP_%change"]); record(f"tradeoff_{r.from_t:.2f}_to_{r.to_t:.2f}_EO_pct", r["EO_%change"])
    record(f"tradeoff_{r.from_t:.2f}_to_{r.to_t:.2f}_PP_pct", r["PP_%change"]); record(f"tradeoff_{r.from_t:.2f}_to_{r.to_t:.2f}_cost", r.tradeoff_cost_EO_loss_per_DP_gain)

# ------------------------------------------------------------------------------
# CELL 15 · Results export (CSV files + Excel workbook)
# ------------------------------------------------------------------------------
model_coefs = pd.concat({"OLS": ols_table.drop(columns="meaning"), "Logistic": logit_table.drop(columns="meaning")}, axis=1)
model_coefs.insert(0, "meaning", [label(i) for i in model_coefs.index]); model_coefs.to_csv(f"{OUT_DIR}/model_coefficients.csv")
final_summary = pd.DataFrame({"item": list(SUMMARY.keys()), "value": [json.dumps(v) if isinstance(v, (list, dict)) else v for v in SUMMARY.values()]})
final_summary.to_csv(f"{OUT_DIR}/final_results_summary.csv", index=False)
show("final_results_summary.csv", final_summary)
with pd.ExcelWriter(f"{OUT_DIR}/CSET485_German_Credit_Results.xlsx", engine="openpyxl") as xw:
    for name, tbl, idx in [("dataset_summary", dataset_summary, False), ("missing_values", missing_table, True), ("outliers", outlier_table, False),
                           ("train_test_split", split_table, True), ("ols_coefficients", ols_table, True), ("logit_coefficients", logit_table, True),
                           ("model_performance", performance, False), ("part1_agreement_matrix", agreement, True), ("part1_mismatch", mismatch_stats, False),
                           ("subgroup_results", subgroup_results, False), ("subgroup_all_predictors", all_predictors, False),
                           ("omitted_variable", ovb_focus, True), ("omitted_variable_all", ovb_full_table, True),
                           ("adversarial_summary", adv_summary, False), ("adversarial_rows", adversarial_out, True), ("adversarial_demographics", demo_table, True),
                           ("adversarial_characteristics", overrep, False), ("fairness_DP_EO_PP", fair_table, False), ("fairness_group_rates", group_stats, False),
                           ("fairness_sex_supplementary", fair_sex.round(4), False), ("tradeoff_consecutive", consecutive, False), ("tradeoff_selected", selected, False),
                           ("final_summary", final_summary, False)]:
        tbl.to_excel(xw, sheet_name=name[:31], index=idx)
print("\nFiles written:")
for root, _, files in os.walk(OUT_DIR):
    for f in sorted(files):
        print("  ", os.path.join(root, f))

# ------------------------------------------------------------------------------
# CELL 16 · SUBMISSION REQUIREMENTS CHECK
# ------------------------------------------------------------------------------
required_files = ["final_results_summary.csv", "fairness_metrics.csv", "model_coefficients.csv", "subgroup_results.csv",
                  "omitted_variable_results.csv", "adversarial_subset.csv", "CSET485_German_Credit_Results.xlsx"]
checks = [
    ("Dataset correctly selected (last digit 4 → German Credit, 1000×21, 700/300)", len(df) == 1000 and df.target.value_counts().to_dict() == {1: 700, 2: 300}),
    ("random_state = 244", RANDOM_STATE == 244 == int(ROLL_NUMBER[-3:])),
    ("70/30 split", abs(len(Xte) / (len(Xtr) + len(Xte)) - 0.30) < 0.005),
    ("Missing values handled", int(missing_after.sum()) == 0),
    ("Categorical encoding completed", all(any(c.startswith(v + "_") for c in X.columns) for v in CATEGORICAL)),
    ("Scaling completed", np.allclose(X_train_s[NUMERIC].mean(), 0, atol=1e-8) and np.allclose(X_train_s[NUMERIC].std(ddof=0), 1, atol=1e-6)),
    ("Outlier analysis completed", len(outlier_table) >= 3 and "log_amount" in df.columns),
    ("OLS completed", hasattr(ols, "params") and len(pred_A) == len(Xte)),
    ("Logistic Regression completed", logit.mle_retvals["converged"] and len(prob_B) == len(Xte)),
    ("Median threshold completed", np.isclose(class_A.mean(), 0.5, atol=0.01)),
    ("0.5 logistic threshold completed", (class_B == (prob_B >= 0.5).astype(int)).all()),
    ("Confusion matrix completed", agreement.values.sum() == len(Xte)),
    ("Mismatch percentage completed", "part1_% mismatched" in SUMMARY),
    ("Subgroup analysis completed", len(subgroup_results) == 3 and subgroup_results.N.sum() == 2 * len(df)),
    ("Omitted-variable analysis completed", len(ovb) >= 4 and logit_reduced.mle_retvals["converged"]),
    ("Four coefficient changes calculated", ovb_focus["pct_change"].notna().sum() >= 4),
    ("Adversarial subset completed", 0 < len(adversarial) <= 50 and len(adversarial) == min(50, n_conf)),
    ("Fairness sweep completed", list(np.round(fair.threshold, 2)) == list(THRESHOLDS)),
    ("DP calculated", "DP" in fair.columns and fair.DP.notna().any()),
    ("EO calculated", "EO" in fair.columns and fair.EO.notna().any()),
    ("PP calculated", "PP" in fair.columns and fair.PP.notna().any()),
    ("Fairness chart completed", os.path.exists(f"{FIG_DIR}/fig06_fairness_tradeoff_DP_vs_EO.png")),
    ("Required output files generated", all(os.path.exists(f"{OUT_DIR}/{f}") for f in required_files)),
]
print("\n" + "=" * 100 + "\nSUBMISSION REQUIREMENTS CHECK\n" + "=" * 100)
for name, ok in checks:
    print(f"{'PASS' if ok else 'FAIL'} — {name}")
print(f"\n{sum(ok for _, ok in checks)} / {len(checks)} checks passed.")
```
